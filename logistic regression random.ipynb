```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
np.random.seed(0)
Study=np.random.randint(1,10,1000)
Sleep=np.random.randint(4,11,1000)
Marks=Study*8 + Sleep*2+np.random.randint(-5,5,1000)
Data = pd.DataFrame({
    'Study':Study,
    'Sleep':Sleep,
    'Marks':Marks
})
x = Data[['Study','Sleep']]
y= Data['Marks']

model = LinearRegression()
model.fit(x,y)

prediction = model.predict([[5,6]])
print(f"Prediction :{prediction}")

prediction_all = model.predict(Data[['Study','Sleep']])

plt.scatter(x['Study'],y,color='green',label='Actual Data')
plt.plot(Data['Study'],prediction_all,color='Red',label='REgression Line')
plt.scatter(5,prediction,color='yellow',s=20,label='User input')
plt.grid(True)
plt.legend()
plt.show()

importance = pd.DataFrame({
    "Features":x.columns,
    "Cofficient":model.coef_
})
print(importance)
```
