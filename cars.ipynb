```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df= pd.read_csv("cars_dataset.csv")
plt.hist(df['ECC'], bins=20, color ='red', edgecolor ='black' )
plt.title("Histogram of Engine Capacity")
plt.show()
```

```python
plt.bar(df["Customer_ID"],df["Price"],color ='red')
plt.xlabel("Customer Id")
plt.ylabel("Price")
plt.title("Bar graph of Customer Id and Price")
plt.show()
```

```python
plt.scatter(df["Customer_ID"],df["Price"],color = 'Purple', marker = '*')
plt.xlabel("Customer Id")
plt.ylabel("Price")
plt.title("Scatter of Customer Id and Price")
plt.show()
```

```python
sns.boxplot(df["Price"],color = 'Purple')
plt.title("Box plot of Price")
plt.show()
```
