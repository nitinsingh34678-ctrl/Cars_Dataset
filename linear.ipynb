```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
n = int(input("Enter NUmber of Data : "))
hours =[]
marks =[]
for i in range(n):
    hour= int(input("Enter hours:")).split()
    mark = int(input("Enter Marks : ")).split()
    hours.append(hour)
    marks.append(mark)

data = pd.Dataframe({
    'Hours':hours,
    'Marks':marks
})
x = data[['Hours']]
y = data['Marks']
```

```python
model = LinearRegression()
model.fit(x,y)
```

```python
user_hours=int(input("Enter user hours :"))
prediction = model.predict([[user_hours]])
```

```python
print("fpredicted marks for{user_hours}hours of Study:{prediction[0]:.2f}")
```

```python
predicted_marks = model.predict(x)
```

```python
plt.scatter(x,y,color='green',label='actual Values')
plt.plot(x,predicted_marks,color='red',label='Regression Line')
plt.scatter(user_hours,predicted_marks,color='purple',s=20,label='Prediction')

plt.xlabel("hours")
plt.ylabel("Marks")
plt.legend()
plt.show()
```
